## 과제에 사용할 클래스 선택

위의 데이터 다운로드 셀을 먼저 실행하세요.

- 학습: 정상 클래스 `2 (Pullover)`만 사용합니다.
- 평가: `0 (T-shirt/top)`, `2 (Pullover)`, `4 (Coat)`, `6 (Shirt)`를 사용합니다.
- `Subset`은 원본 데이터와 라벨을 유지하면서 필요한 샘플만 선택합니다. 이후 학습에는 `train_set`, 평가에는 `test_set`을 사용합니다.
- 위에서 설정한 `ToTensor()`는 샘플을 꺼낼 때 이미지를 `[1, 28, 28]` 텐서와 `[0, 1]` 범위로 변환합니다.
- 테스트 라벨은 클래스 선택과 평가에만 사용하며, 학습이나 이상 점수 계산에는 사용하지 않습니다.

```python
import torch
from torch.utils.data import Subset

normal_class = 2
test_classes = torch.tensor([0, 2, 4, 6])

train_indices = torch.where(train_full.targets == normal_class)[0].tolist()
test_indices = torch.where(torch.isin(test_full.targets, test_classes))[0].tolist()

train_set = Subset(train_full, train_indices)
test_set = Subset(test_full, test_indices)

print("선택한 학습 데이터:", len(train_set))
print("학습 라벨:", train_full.targets[train_indices].unique().tolist())
print("선택한 테스트 데이터:", len(test_set))
labels, counts = test_full.targets[test_indices].unique(return_counts=True)
print("테스트 클래스별 개수:", dict(zip(labels.tolist(), counts.tolist())))

image, label = train_set[0]
print("이미지 형태:", tuple(image.shape), "자료형:", image.dtype)
print("픽셀 범위:", image.min().item(), image.max().item())
```
